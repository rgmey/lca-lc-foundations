## No memory

```python
from langchain_openrouter import ChatOpenRouter
model = ChatOpenRouter(model="openai/gpt-5-nano")   # or anthropic/claude-3.5-haiku, etc.
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=model
    )
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is LilSMD and my favourite colour is green")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

## Memory

```python
from langgraph.checkpoint.memory import InMemorySaver  

agent = create_agent(
    model=model,
    checkpointer=InMemorySaver(),  
    )

# agent = create_agent(
#     "gpt-5-nano",
#     checkpointer=InMemorySaver(),  
# )
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
```

```python
pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour? say it in one word only!")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
```
